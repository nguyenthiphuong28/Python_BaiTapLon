```python
import pandas as pd

df = pd.read_csv("dataset.csv")
```

```python
# Đếm số bài theo từng thể loại

genre_count = df.groupby("track_genre") \
                .size() \
                .reset_index(name="so_luong") \
                .sort_values(by="so_luong", ascending=False)

print(genre_count.head(20))
```

```python
#Top 10 bài hát có độ phổ biến cao nhất

top10 = df[["track_name", "artists", "popularity"]] \
    .sort_values(by="popularity", ascending=False) \
    .head(10)

print(top10)
```

```python
#Thống kê thời lượng & độ phổ biến theo thể loại
genre_stats = df.groupby("track_genre").agg(
    so_luong=("track_name", "count"),
    thoi_luong_tb=("duration_ms", "mean"),
    popularity_tb=("popularity", "mean")
).reset_index()

genre_stats = genre_stats.sort_values(by="popularity_tb", ascending=False)

print(genre_stats.head(20))
```

```python
#Thống kê số lượng bài hát & độ phổ biến theo nghệ sĩ

df_artists = df.assign(artist=df["artists"].str.split(";")).explode("artist")

artist_stats = df_artists.groupby("artist").agg(
    so_luong=("track_name", "count"),
    popularity_tb=("popularity", "mean")
).reset_index()

artist_stats = artist_stats.sort_values(by="so_luong", ascending=False)

print(artist_stats.head(20))
```

```python
#Thể loại có energy & danceability trung bình cao nhất

genre_energy = df.groupby("track_genre").agg(
    energy_tb=("energy", "mean"),
    danceability_tb=("danceability", "mean")
).reset_index()

genre_energy = genre_energy.sort_values(
    by=["energy_tb", "danceability_tb"],
    ascending=False
)

print(genre_energy.head(20))
```
